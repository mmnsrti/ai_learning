```python
import numpy as np

def perceptron_fit(X, y, lr=0.1, epochs=25):
    w = np.zeros(X.shape[1], dtype=float)
    b = 0.0
    for _ in range(epochs):
        for xi, yi in zip(X, y):
            yhat = 1 if np.dot(w, xi) + b >= 0 else 0
            err = yi - yhat
            if err != 0:
                w += lr * err * xi
                b += lr * err
    return w, b

# AND gate
X = np.array([[0,0],[0,1],[1,0],[1,1]], float)
y = np.array([0,0,0,1])
w, b = perceptron_fit(X, y)
def predict(X): return (X @ w + b >= 0).astype(int)

print("w,b =", w, b)
print("pred:", predict(X))  # should be [0 0 0 1]
```

```python
from sklearn.linear_model import Perceptron
clf = Perceptron(max_iter=1000, eta0=0.1, tol=1e-3)
clf.fit(X, y)
clf.predict(X)

```
