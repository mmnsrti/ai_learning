```python
def check_and_change_name(name):

    if name == 'Sorush':
        return 'Shayan'
    else:
        return name
```

```python

def process_name_list(names_list) :
    updated_names = []
    for name in names_list:
        new_name = check_and_change_name(name)
        updated_names.append(new_name)
    return updated_names
```

```python
original_names = [
    "Sara",
    "Sorush",
    "Ali",
    "Sorush",
    "Mona",
    "sorush",  
    "Shayan"
]
```

```python
corrected_names = process_name_list(original_names)
```

```python
print(f"Original List: {original_names}")
print(f"Corrected List: {corrected_names}")
```
