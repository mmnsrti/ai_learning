## Background: What is TF-IDF?

**TF-IDF** reflects how important a word is to a document in a corpus.

$$TF(t, d) = \frac{\text{count of term } t \text{ in doc } d}{\text{total terms in doc } d}$$

$$IDF(t) = \log\left(\frac{\text{Total documents}}{\text{Documents containing } t}\right)$$

$$TF\text{-}IDF = TF \times IDF$$
